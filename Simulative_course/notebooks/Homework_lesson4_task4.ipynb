Первичные действия:

- Прочитайте содержимое файла без участия _Pandas_ в переменную _f_. 

- Сформируйте список _lst_, который будет состоять из списков, где каждый список представляет собой строку файла. Длина каждого списка равняется количеству столбцов в файле.

- Если в названиях столбцов присутсвуют лишние символы - уберите их. 

- Если вместе с данными загрузились полностью пустые строки - уберите их. 

- В конце создайте датафрейм Pandas и назовите его _df_. Названия столбцов и содержимое возьмите из созданного ранее списка _lst_.

```python
import pandas as pd
```

```python
text = open('D:/GitHub_projects/Simulative_course/data/itresume-coderun.csv', 'r')
f = text.read()
lst = [line.replace('"', '').split(sep=',') for line in f.splitlines() if line != ''][:-1]
lst
```

```python
df = pd.DataFrame(data=lst[1:], columns=lst[0])
df
```
